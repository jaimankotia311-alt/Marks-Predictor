```python
import numpy as np

np.random.seed(42)

hours_studied = np.random.randint(1, 19 ,300)

exam_score = (
    28
    + (hours_studied * 4)
    + np.random.normal(0,5,300)
)

exam_score = np.clip(exam_score,30 , 100).round().astype(int)

data = {
    "Hours Studied" : hours_studied.tolist(),
    "Exam Score"    : exam_score.tolist()
}
print(data)
```

```python
import pandas as pd
```

```python
y = df["Exam Score"]
y
```

```python
print(x.head())                               # input column , independent
print(y.head())                               # output column , dependent
print("x dimension = ",x.ndim)
print("x type= ",type(x))
print("y dimension = ",y.ndim)
print("y type= ",type(y))


```

```python
# split
# training  = 80 % x_train , 80% y_train
# testing   = 20 % x_test  , 20% y_test
```

```python
# train_test_split
# sklearn 
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train , y_test = train_test_split(x,y,test_size = 0.2)
```

```python
print("x_train shape =", x_train.shape)
print("x_test shape =", x_test.shape)
print("y_train shape =", y_train.shape)
print("y_test shape =", y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(x_train ,y_train)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
y_pred = model.predict(x_test)
```

```python

y_test.head()
```

```python
y_test.shape
```

```python
accuracy = r2_score(y_test, y_pred) 
accuracy *100
```

```python
# training
r2_score(y_train,model.predict(x_train))*100
```

```python
plt.scatter(x_train, y_train, color = "red" ,label = "Training Data")
plt.scatter(x_test, y_test, color = "green" ,label = "Testing Data")
plt.plot(x_test, y_pred, color = "black" ,label = "BFT")
plt.legend()
plt.title("BFT for Trainig Data")
```

```python
# bias/error
# actual output - predicted output

pd.DataFrame(
    {"Actual Output" : y_test,
    "Predicted Output" : y_pred,
    "Bias/Error" : y_test - y_pred,
    "Error in % ": (((y_test - y_pred)/y_test)*100 ).round(2)
    }
)
```

```python
plt.scatter(x_train, y_train, color = "red", label = "Training Data")
plt.scatter(x_test, y_test, color = "green", label = "Testing Data")
plt.plot(x_train, model.predict(x_train), color = "blue", label = "BFT")
plt.legend()
plt.title("BFT for Training Data")
plt.show()
```
